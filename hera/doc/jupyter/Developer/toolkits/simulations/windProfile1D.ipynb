### getSpatialWind()

For calculating the Wind Profile in specific regions, we first specify such region and calculate the corresponding spatial wind values.  
We use it using the **getSpatialWind()** function.  
Arguments:  
- minlat: Minimum value of latitude coodinate (X axis).
- minlat: Minimum value of longitude coodinate (Y axis).
- maxlon: Maximum value of latitude coodinate (X axis).
- maxlat: Maximum value of longitude coodinate (Y axis).
- IMS_TOKEN: IMS token for connection IMS API.
- dxdy: Spatial resolution of the output land cover map. Defines the step size for the grid points in both the x (longitude) and y (latitude) directions within the specified bounding box.
- inputCRS: The ESPRG of the coordinates.
- landcover_DataSource: The name of the landcover DataSource to use.

```python
minx = 179000
miny = 667762
maxx = 182290
maxy = 670516
f = open(f'token.json')
token = json.load(f)

xr_dataset,stations_with_data = wind_tk.getSpatialWind(minx,miny,maxx,maxy,token,dxdy=500,inputCRS=ITM)
```

The function returns two values. The first one is the corresponding xarray of the specified region. The second is the stations in the region (currently using all wind stations in Israel).

### getWindProfile()  
Now, we can caluclate the wind profile of each cell in each specified height.
We do this using **getWindProfile()**.  
The output will be a dataframe with lat,lon,height,u,v and U_Z columns.  
Arguments:  
- xarray: Xarray of the region.
- height: The Maximum height value to calculate U and V.
- dz: Resolution for height axis.

## Plot Wind Directions  
We can draw the wind directions in each lat lon coordinate in our xarray mesh.  
We do this using the Presentation layer of the toolkit, using the **plotWindDirections()** function.

First, we plot the satellite image of our mesh using the tile toolkit:

```python
from hera.measurements.GIS.utils import WSG84,ITM,convertCRS
from hera import toolkitHome
import matplotlib.pyplot as plt

minx = 179000
miny = 667762
maxx = 182290
maxy = 670516

tiles_tk = toolkitHome.getToolkit(toolkitHome.GIS_TILES)
TA = dict(minx=minx,miny=miny,maxx=maxx,maxy=maxy,zoomlevel=15,tileServer=tiles_tk.getDataSourceList()[0],inputCRS=ITM, outputCRS=WSG84)
img = tiles_tk.getImageFromCorners(**TA)
fig,ax = plt.subplots(1,1,figsize=(28,28))
mesh_satellite_image = tiles_tk.presentation.plot(img,inputCRS=WSG84,outputCRS=ITM,ax=ax)
```

Arguments:  
- plot: The plot image of the mesh.
- landcover: The Xarray of the mesh.
- figsize: Figure size. Default is (28,28).
- color: Arrow color. Default is set to 'black'.
- arrow_length: Arrow length. Default is 10.
- width: Arrow width. Default is 10.
- head_width: Arrow head width. Default is 50.
- head_length: Arrow head length. Default is 70.
- show_photo: Boolean argument to show image or not. Default=True. If is set to False, returns the figure.

```python
wind_tk.presentation.plotWindDirections(mesh_satellite_image,xr_dataset)
```
