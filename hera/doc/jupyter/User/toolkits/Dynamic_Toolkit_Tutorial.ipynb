# Dynamic Toolkits – User Toutorial

dynamic toolkits are just like regiular toolkits, only that they are stored in the database.
you simply use them with 
tk = toolkitHome.getToolkit(toolkitName=toolkit_name).
this command will work for all toolkits and experiments.

## adding a dynamic toolkit:

### 1) Create an exmple toolkit 


```python
import os

# Base directory where toolkits live
base_dir = os.path.join(os.getcwd(), "toolkits")

toolkit_name = "myExampleToolkit"
toolkit_dir = os.path.join(base_dir, toolkit_name)

# Create directory if needed
os.makedirs(toolkit_dir, exist_ok=True)

# Toolkit file path
toolkit_file = os.path.join(toolkit_dir, f"{toolkit_name}.py")

# Write the toolkit class
with open(toolkit_file, "w") as f:
    f.write(
        f"""
class {toolkit_name}:
    def __init__(self, projectName=None, filesDirectory=None, **kwargs):
        self.projectName = projectName
        self.filesDirectory = filesDirectory

    def hello(self):
        return f"Hello from {{self.__class__.__name__}}"
"""
    )

print(f"Toolkit created at: {toolkit_file}")

```


### 2) add the toolkit to the database
 
Run the following command in your terminal:

<p style="background:black">
<code style="background:black;color:white">>> hera-project addToolkit &lttoolkit_name&gt &toolkit_path&gt [--params '{"param1": 42}'] [ --version "0,0,1"] [ --overwrite ]
</code>
</p> 

Parameters:

toolkit_name — Name of the toolkit

toolkit_path — Path to toolkit directory

--params — Optional JSON parameters

--version — Toolkit version default: (0,0,1)

--overwrite — Overwrite existing toolkit if exists

### 3) Retrieve the toolkit


```python
from hera import toolkitHome

tk = toolkitHome.getToolkit(
    toolkitName=toolkit_name,
    version = (1,0,0)
    # filesDirectory="/tmp/toolkit_files",
    # someArg=123,
)

print("Toolkit retrieved:", tk)
```

### 4) check the retrived toolkit class


```python
print(tk.hello())
```
