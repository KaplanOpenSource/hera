# Hera Repository Attacher – Quick Guide

This notebook shows how to:
1. Install prerequisites
2. (Optionally) create sample users
3. Use `hera_repo_cli.py add` to attach a repo to all users
4. Verify with `hera_repo_cli.py check`

> Adapt host/port/credentials to your environment.

## 0. Files
- `hera_repo_cli.py` – CLI wrapper
- `add_repo_to_all_users.py` – the underlying script (ensure it is in the same folder or adjust `--script-path`)

```python
# 1) Install prerequisites (run in your environment)
!pip install pymongo click
```

## 2) (Optional) Create sample users in DB `hera`
If you want to test locally. Requires a MongoDB user with rights on `hera`.

```python
%%bash
mongosh \
  --host 127.0.0.1 --port 27017 \
  --username "<mongo-user>" --password "<mongo-pass>" --authenticationDatabase "dbhera" <<'EOS'
use hera
db.users.insertMany([
  {_id: "u1", username: "alice",  repos: []},
  {_id: "u2", username: "bob",    repos: []},
  {_id: "u3", username: "charlie",repos: []}
])
db.users.find().pretty()
EOS
```

## 3) Run: attach repo to all users (scan DBs by prefix)
Adjust `--repo-name`, `--repo-path`, and credentials.

```python
%%bash
python hera_repo_cli.py add \
  --repo-name IMS \
  --repo-path /opt/hera/toolkits/IMS \
  --username <mongo-user> \
  --password <mongo-pass> \
  --auth-db dbhera \
  --db-prefix hera \
  --verbose
```

## 4) Verify results

```python
%%bash
python hera_repo_cli.py check \
  --username <mongo-user> \
  --password <mongo-pass> \
  --auth-db dbhera
```

## Notes
- The operation is idempotent; re-running `add` will typically modify 0 users after the first run.
- For a single DB (no scanning), you can use the underlying script directly with `--db-list <db1,db2>`.
